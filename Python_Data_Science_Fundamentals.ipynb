# 1. NumPy Fundamentals

NumPy is a Python library used for numerical computing and working with arrays.

```python
# Creating a NumPy array

marks = np.array([78, 65, 92, 55, 88])

print("Marks:", marks)
print("Shape:", marks.shape)
print("Size:", marks.size)
print("Dimensions:", marks.ndim)
print("Data type:", marks.dtype)
```

```python
# NumPy calculations

print("Total marks:", np.sum(marks))
print("Average marks:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))
print("Standard deviation:", np.std(marks))
```

```python
# NumPy indexing and slicing

print("First mark:", marks[0])
print("Second mark:", marks[1])
print("Last mark:", marks[-1])

print("First three marks:", marks[:3])
print("Last two marks:", marks[-2:])
```

```python
# Creating a 2D NumPy array

student_marks = np.array([
    [78, 82, 75],
    [65, 70, 68],
    [92, 89, 94]
])

print("Student marks:")
print(student_marks)

print("\nShape:", student_marks.shape)
print("Rows:", student_marks.shape[0])
print("Columns:", student_marks.shape[1])
```

# 2. Pandas Fundamentals

Pandas is a Python library used for data manipulation and data analysis.
It provides useful structures such as Series and DataFrames.

```python
# Creating a student DataFrame

data = {
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Arjun"],
    "Math": [78, 65, 92, 55, 88],
    "Science": [82, 70, 89, 60, 91],
    "English": [75, 68, 94, 58, 85]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Calculate total and average marks

df["Total"] = df["Math"] + df["Science"] + df["English"]
df["Average"] = df["Total"] / 3

print(df)
```

```python
# Determine Pass or Fail

df["Result"] = np.where(df["Average"] >= 40, "Pass", "Fail")

print(df)
```

```python
# Find highest and lowest performers

highest = df.loc[df["Average"].idxmax()]
lowest = df.loc[df["Average"].idxmin()]

print("Highest Performer:")
print(highest)

print("\nLowest Performer:")
print(lowest)
```

```python
# Students with an average of 75 or above

high_performers = df[df["Average"] >= 75]

print("Students with average marks of 75 or above:")
print(high_performers)
```

```python
# Sort students by average marks

ranked_students = df.sort_values(by="Average", ascending=False)

print("Students ranked by average marks:")
print(ranked_students)
```

```python
# Calculate average marks for each subject

math_average = df["Math"].mean()
science_average = df["Science"].mean()
english_average = df["English"].mean()

print("Math average:", math_average)
print("Science average:", science_average)
print("English average:", english_average)
```

```python
# Display the final dataset

print("Final Student Performance Dataset:")
print(df)
```

# 3. Data Visualization

Matplotlib is a Python library used to create charts and graphs.
In this project, visualization helps us understand student performance.

```python
# Bar chart of student average marks

plt.figure(figsize=(10, 5))

plt.bar(df["Name"], df["Average"])

plt.title("Student Average Marks")
plt.xlabel("Student")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Subject-wise average marks

subjects = ["Math", "Science", "English"]
averages = [
    df["Math"].mean(),
    df["Science"].mean(),
    df["English"].mean()
]

plt.figure(figsize=(8, 5))

plt.bar(subjects, averages)

plt.title("Subject-wise Average Marks")
plt.xlabel("Subject")
plt.ylabel("Average Marks")

plt.tight_layout()
plt.show()
```

```python
# Line chart of student average marks

plt.figure(figsize=(10, 5))

plt.plot(df["Name"], df["Average"], marker="o")

plt.title("Student Performance")
plt.xlabel("Student")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
