## 1. Import Libraries

along with Pandas and NumPy for data handling.

## 2. Create the Dataset

We create a simple student performance dataset containing
student names, study hours, scores, and attendance.

```python
data = {
    "Student": ["Rahul", "Priya", "Aman", "Neha", "Rohit",
                "Sneha", "Arjun", "Kavya", "Vivek", "Anjali"],
    "Hours_Studied": [2, 3, 4, 5, 6, 7, 8, 5, 9, 10],
    "Score": [40, 48, 55, 62, 70, 76, 82, 65, 90, 95],
    "Attendance": [70, 75, 78, 82, 85, 88, 90, 84, 94, 96]
}

df = pd.DataFrame(data)

df
```

## 3. Explore the Dataset

Before creating charts, we examine the structure and
summary statistics of the dataset.

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe()
```

## 4. Bar Chart

A bar chart is used to compare the scores of different students.

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Student"], df["Score"])

plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Student Scores")

plt.xticks(rotation=45)

plt.show()
```

## 5. Line Chart

A line chart is used to show trends.

Here, we visualize the relationship between hours studied
and student scores.

```python
plt.figure(figsize=(8, 5))

plt.plot(df["Hours_Studied"], df["Score"], marker="o")

plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.title("Study Hours vs Score")

plt.grid(True)

plt.show()
```

## 6. Pie Chart

A pie chart is used to show the proportion of different categories.

Here, we categorize students based on their scores.

```python
def score_category(score):
    if score >= 80:
        return "Excellent"
    elif score >= 60:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance"] = df["Score"].apply(score_category)

performance_counts = df["Performance"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    performance_counts,
    labels=performance_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Performance Distribution")

plt.show()
```

## 7. Histogram

A histogram is used to understand the distribution of numerical data.

Here, we visualize the distribution of student scores.

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Score"], bins=5, edgecolor="black")

plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Scores")

plt.show()
```

## 8. Scatter Plot

A scatter plot is used to identify relationships and patterns
between two numerical variables.

Here, we compare hours studied with student scores.

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Hours_Studied"],
    df["Score"],
    s=100
)

plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.title("Hours Studied vs Student Score")

plt.grid(True)

plt.show()
```

## 10. Key Insights

From the visualizations, we can identify the following patterns:

- Students who studied more hours generally achieved higher scores.
- The line chart shows an upward trend between study hours and scores.
- The bar chart makes it easy to compare individual student scores.
- The histogram shows the distribution of student scores.
- The pie chart shows the proportion of students in different performance categories.
- The scatter plot shows a positive relationship between hours studied and scores.
- Attendance also generally increases for students with higher scores.

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    df[["Hours_Studied", "Score", "Attendance"]].corr(),
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()
```
